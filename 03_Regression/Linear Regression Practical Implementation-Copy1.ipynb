```python
import pandas as pd
```

```python
from sklearn.datasets import fetch_california_housing
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=fetch_california_housing()
```

```python
df
```

```python
dataset = pd.DataFrame(df.data)
```

```python
dataset
```

```python
dataset.columns=df.feature_names
```

```python
dataset.head(10)
```

```python
## Independent features and dependent features
X=dataset
y=df.target #output
```

```python
y
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test= train_test_split(X,y,test_size =0.25,random_state =42)
```

```python
X_train
```

```python
## standardizing the dataset
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
```

```python
X_test=scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
##cross validation
from sklearn.model_selection import cross_val_score
```

```python
regression=LinearRegression()
regression.fit(X_train,Y_train)
```

```python
mse=cross_val_score(regression,X_train,Y_train,scoring='neg_mean_squared_error',cv=100) #check_model_error
```

```python
np.mean(mse)    #Iska matlab hai ki data ko 100 folds (tukdo) mein divide kiya gaya hai. Model 100 alag-alag baar train aur test hota hai, aur aakhri mein 100 errors ki list return karta hai (jiska average aapne line 34 mein np.mean(mse) karke -0.52 nikala hai).
```

```python
##prediction 
reg_pred=regression.predict(X_test)
```
